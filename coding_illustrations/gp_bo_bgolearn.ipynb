```python
import numpy as np
import pandas as pd
import Bgolearn.BGOsampling as BGOS

# ---------- Test objective (maximisation) ----------
def f_true(X):
    # X: shape (N, 2)
    x1 = X[:, 0]
    x2 = X[:, 1]
    return np.sin(5 * np.pi * x1) * np.cos(5 * np.pi * x2) + \
           0.5 * np.cos(10 * np.pi * x1) * np.sin(10 * np.pi * x2)

# ---------- Initialisation ----------
n_init = 5
X = np.random.rand(n_init, 2)
y = f_true(X)

# Convert to DataFrame (Bgolearn style)
data = pd.DataFrame(np.hstack([X, y.reshape(-1, 1)]))

# ---------- Virtual sampling space ----------
grid_size = 100
x1 = np.linspace(0, 1, grid_size)
x2 = np.linspace(0, 1, grid_size)
X1, X2 = np.meshgrid(x1, x2)
vs = np.vstack([X1.ravel(), X2.ravel()]).T   # (N_grid, 2)

# ---------- BGO object ----------
Bgolearn = BGOS.Bgolearn()

# ---------- BO loop ----------
for t in range(20):

    # ----- Fit surrogate model -----
    model = Bgolearn.fit(
        data_matrix=data.iloc[:, :-1],
        Measured_response=data.iloc[:, -1],
        virtual_samples=vs
    )

    # ----- Acquisition (Expected Improvement) -----
    _, x_next = model.EI()   # directly use Bgolearn's recommended point
    x_next = np.array(x_next).reshape(1, -1)

    # ----- Evaluate true function -----
    y_next = f_true(x_next).item()

    # ----- Update dataset -----
    new_row = np.hstack([x_next.flatten(), y_next])
    data.loc[len(data)] = new_row

    # ----- Logging -----
    best_y = data.iloc[:, -1].max()
    print(f"Iter {t+1}: selected x = [{x_next[0,0]:.4f}, {x_next[0,1]:.4f}], "
          f"y = {y_next:.4f}, best so far = {best_y:.4f}")

# ---------- Final recommendation ----------
best_idx = data.iloc[:, -1].idxmax()
x_best = data.iloc[best_idx, :-1].values
y_best = data.iloc[best_idx, -1]

print(f"\nBest found: x = [{x_best[0]:.4f}, {x_best[1]:.4f}], y = {y_best:.4f}")
```
