The training data are stored in `./data/train.xlsx`, and the search-space data are stored in `./data/visual_samples.xlsx`.

The dataset includes four input features, namely Sn, Bi, In, and Ti, together with one target variable, T.

The objective is to determine the combination of Sn, Bi, In, and Ti that maximizes T.

```python
# import training data
import pandas as pd
data = pd.read_excel('./data/train.xlsx')
print(data)
```

```python
# import search space data
import pandas as pd
vs = pd.read_excel('./data/visual_samples.xlsx')
print(vs)
```

```python
import Bgolearn.BGOsampling as BGOS
Bgolearn = BGOS.Bgolearn() 

Mymodel = Bgolearn.fit(data_matrix = data.iloc[:,:-1], Measured_response = data.iloc[:,-1], virtual_samples = vs,)
Mymodel.EI()
```

```python
import Bgolearn.BGOsampling as BGOS
Bgolearn = BGOS.Bgolearn() 

Mymodel = Bgolearn.fit(data_matrix = data.iloc[:,:-1], Measured_response = data.iloc[:,-1], virtual_samples = vs)
Mymodel.EI_log()
```

```python
import Bgolearn.BGOsampling as BGOS
Bgolearn = BGOS.Bgolearn() 
Mymodel = Bgolearn.fit(data_matrix = data.iloc[:,:-1], Measured_response = data.iloc[:,-1], virtual_samples = vs)
Mymodel.UCB()
```
